# Question
# what is the joint probability of pulling a card from a deck of cards that is "black" AND "face"?


# Simulation Description
 1. The random process that I will be simulating is the act of randonly drawing a card, without replacement from a standard 52 deck cards. 
 2. The random process in my simulation is the act of drawing random cards  and then another. The possible outcomes of one trial is the 2 cards found in a 52-card deck. 
3. The compound event I will be measuring is that wheter two events, one card being a "black" and at the same time being a "face" card can happen?  

# Revised  Description
1. A card being "black" is A and a card being "face" is B. 
2. My question is asking for A and B, which is a joint question, as I am looking for a card that is simultaneusly "black" AND "face". 
3. A and B are not mutually exclusive, as getting a black card or a getting a face card does not prevent me from getting a card that is the opposite. 
4. A and B are dependent, because if I get a card that is "black" or if I get a card that is "face", it decreases the chances of my next card being black or face. therefore, my next card being black and face is decreased. 

```python
import pandas as pd

# 1. Define the components of a standard deck of cards
suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
ranks = ["2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King", "Ace"]
```

```python
# 2. Build the deck DataFrame
deck = []
for suit in suits:
    for rank in ranks:
        # Hearts and Diamonds are Red; Clubs and Spades are Black
        color = "Red" if suit in ["Hearts", "Diamonds"] else "Black"

        # Jacks, Queens, and Kings are Face cards
        is_face = rank in ["Jack", "Queen", "King"]

        deck.append(
            {"suit": suit, "rank": rank, "color": color, "is_face": is_face}
        )

df = pd.DataFrame(deck)
```

```python
# 3. Filter for cards that are BOTH "Black" AND "Face"
black_and_face = df[(df["color"] == "Black") & (df["is_face"] == True)]
```

```python
# 4. Calculate the probability
total_cards = len(df)
matching_cards = len(black_and_face)
probability = matching_cards / total_cards
```

```python
# 5. Display the results
print(f"Total cards in deck: {total_cards}")
print(f"Number of black face cards: {matching_cards}")
print(f"Joint Probability: {probability:.4f} (or {matching_cards}/{total_cards})")
```
